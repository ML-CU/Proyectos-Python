# Hipotesis Testing P-Value

¿Cual es la probabilidad de obtener este numero de sellos en x lanzamientos, si la moneda es justa?. La respuesta es que la moneda tiene una probabilidad de sacar en un 50% cara y un 50% sello. Lo comentado anteriormente es el valor de P.

* $ P(Datos|H_{0}) $ : Cual es la probabilidad de observar tales datos dado que lo que yo creo es verdad Ho (Hipotesis nula)
* Cual es la probabilidad de observar exactamente 80 caras en 100 lanzamientos si yo creo que la moneda es justa. como es una binomial tenemos como resultado 4.22 a la-10. Aqui partimos de que la moneda es justa y concluimos de que es muy poco probable obtener 80 caras. Entonces hya algo raro en esa moneda que esta dando 80 caras en 100 lanzamientos.

$ P(Datos|H_{0}) $ Entonces ***P es la probabilidad de observar, de obtener esos datos dado que mi hipotesis nula es cierta***

#### Hagamos un experimento, usemos como ejemplo una población normal con media 120 y desviación estandar 9

```python
mu = 120
sd = 9
size = 1341
pop = np.random.normal(loc=mu, scale=sd, size=size)
```

```python
medias = []
trials = 50000
sample_size = 21
for i in range(trials):
    medias.append(np.random.choice(pop, sample_size).mean())
```

```python
plt.hist(medias)
```

No se cuanto vale $\mu$, pero me dicen que por experiencia $\mu = 120$

```python
#hagamos un experimento con una muestra de 21 para ver si realmente mu es 120
np.random.choice(pop, sample_size).mean()
```

Ahora realicemos una prueba de hipotesis
* H0 es lo que piensa el jefe
* Ha es lo que pienso yo
#### hipotesis
$ H_{0}: \mu = 120 $

$ H_{a}: \mu > 120 $

¿Cual es la probabilidad de obtener una media de 117.30 SI REALMENTE la $\mu = 120$?

```python
#calculemos el p-valor de obtener una media muestral mayor a 117.30 si la media poblacional es 120
(np.array(medias) > 117.30).sum()/len(medias)
```

Si el valor de P < 0.05 rechazo la hipotesis nula H0, de lo contrario la acepto.
entonces podriamos decirle al jefe que esta equivocado, porque es muy poco probable que haya obtenido un 117.30 si realmente es 120.

¿Que pasaria si P nos daria 0.01?
podriamos decir que el jefe esta en lo correcto

A este numero P se le llama $ \alpha\ $ que es el nivel de significancia.

Para resumir el valor de P **es la probabilidad de observar mis datos si yo asumo que lo que yo pienso es cierto**
